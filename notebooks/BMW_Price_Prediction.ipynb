```python
def compute_cost(X, y, w, b):
    m = X.shape[0]
    
    predictions = X @ w + b
    errors = predictions - y
    squared_errors = errors ** 2
    
    cost = (1 / (2 * m)) * np.sum(squared_errors)
    
    return cost
```

```python
cost = compute_cost(X_train_scaled, y_train, w, b)

print("Cost initial :", cost)
```

```python
cost_verification = (1 / (2 * len(y_train))) * np.sum(y_train ** 2)

print("Cost vérifié :", cost_verification)
```

```python
def compute_gradient(X, y, w, b):
    m = X.shape[0]
    
    predictions = X @ w + b
    errors = predictions - y
    dj_dw = (1 / m) * (X.T @ errors)
    dj_db = (1 / m) * np.sum(errors)
    return dj_dw, dj_db
```

```python
dj_dw, dj_db = compute_gradient(
    X_train_scaled,
    y_train,
    w,
    b
)

print("Shape de dj_dw :", dj_dw.shape)
print("Valeur de dj_db :", dj_db)
print("Premiers gradients :", dj_dw[:5])
```

```python
alpha = 0.01

print("Learning rate :", alpha)
```

```python
alpha = 0.01

print("Learning rate :", alpha)
```

```python
def gradient_descent(X, y, w, b, alpha, num_iters):
    for i in range(num_iters):
        dj_dw, dj_db = compute_gradient(X, y, w, b)

        w = w - alpha * dj_dw
        b = b - alpha * dj_db

    return w, b
```

```python
w_initial = np.zeros(X_train_scaled.shape[1])
b_initial = 0

alpha = 0.01
num_iters = 1000

print("Nombre de poids :", len(w_initial))
print("Biais initial :", b_initial)
print("Learning rate :", alpha)
print("Nombre d'itérations :", num_iters)
```

```python
w_trained, b_trained = gradient_descent(
    X_train_scaled,
    y_train,
    w_initial,
    b_initial,
    alpha,
    num_iters
)

print("Entraînement terminé !")
print("Biais entraîné :", b_trained)
print("Premiers poids :", w_trained[:5])
```

```python
cost_trained = compute_cost(
    X_train_scaled,
    y_train,
    w_trained,
    b_trained
)

print("Cost après entraînement :", cost_trained)
```

```python
import sys
import numpy as np
import matplotlib.pyplot as plt

# Ajouter le dossier racine du projet au chemin Python
sys.path.append("..")

from src.preprocessing import preprocess_data
from src.cost_function import compute_cost


# ============================================================
# 1. CHARGEMENT ET PRÉPARATION DES DONNÉES
# ============================================================

X_train, X_test, y_train, y_test, scaler, encoder = preprocess_data(
    "../data/bmw.csv"
)

# Convertir y en tableau NumPy
y_train_np = y_train.to_numpy()


# ============================================================
# 2. FONCTION COMPUTE_GRADIENT
# ============================================================

def compute_gradient(X, y, w, b):
    """
    Calcule le gradient par rapport à w et b.
    """

    m = X.shape[0]

    # Prédictions
    predictions = X @ w + b

    # Erreurs
    errors = predictions - y

    # Gradient par rapport aux poids
    dj_dw = (1 / m) * (X.T @ errors)

    # Gradient par rapport au biais
    dj_db = (1 / m) * np.sum(errors)

    return dj_dw, dj_db


# ============================================================
# 3. INITIALISATION
# ============================================================

w = np.zeros(X_train.shape[1])
b = 0

alpha = 0.01
num_iters = 1000

cost_history = []


# ============================================================
# 4. GRADIENT DESCENT
# ============================================================

for i in range(num_iters):

    # Calcul du gradient
    dj_dw, dj_db = compute_gradient(
        X_train,
        y_train_np,
        w,
        b
    )

    # Mise à jour des paramètres
    w = w - alpha * dj_dw
    b = b - alpha * dj_db

    # Calcul du coût
    cost = compute_cost(
        X_train,
        y_train_np,
        w,
        b
    )

    # Sauvegarde du coût
    cost_history.append(cost)


# ============================================================
# 5. AFFICHAGE DES RÉSULTATS
# ============================================================

print("Gradient Descent terminé.")
print(f"Nombre d'itérations : {num_iters}")
print(f"Learning rate : {alpha}")
print(f"Coût initial : {cost_history[0]:,.2f}")
print(f"Coût final : {cost_history[-1]:,.2f}")
print(f"Biais final : {b:,.2f}")


# ============================================================
# 6. GRAPHIQUE DE LA FONCTION DE COÛT
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(cost_history)

plt.title("Évolution de la fonction de coût")
plt.xlabel("Itérations")
plt.ylabel("Coût")

plt.tight_layout()


# ============================================================
# 7. SAUVEGARDE DU GRAPHIQUE
# ============================================================

plt.savefig(
    "../images/cost_function.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
